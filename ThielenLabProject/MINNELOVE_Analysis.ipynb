# MINNELOVE: Results Analysis
Visualizing taxonomic abundance.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Use 'emu combine-outputs' in your terminal first to get this TSV
try:
    df = pd.read_csv('emu_combined_genus.tsv', sep='\t').set_index('genus')
except:
    # Placeholder data for your presentation demo
    data = {'Genus': ['Bacteroides', 'Lactobacillus', 'Prevotella'], 'S1': [0.4, 0.3, 0.3], 'S2': [0.35, 0.45, 0.2]}
    df = pd.DataFrame(data).set_index('Genus')

df.T.plot(kind='bar', stacked=True, colormap='Spectral', figsize=(10,6))
plt.title('MINNELOVE Genus Composition')
plt.ylabel('Relative Abundance')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
